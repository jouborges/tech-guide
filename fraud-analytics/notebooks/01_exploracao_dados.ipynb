```python
fraudes = df[df["Class"] == 1]

quantidade = (fraudes["Amount"] <= 10).sum()

percentual = quantidade / len(fraudes) * 100

print(f"Fraudes de valor até 10: {quantidade}")
print(f"Percentual: {percentual:.2f}%")
```

```python
fraudes = df[df["Class"] == 1]

quantidade = (fraudes["Amount"] <= 10).sum()
percentual = quantidade / len(fraudes) * 100

print(f"Fraudes de valor até 10: {quantidade}")
print(f"Percentual: {percentual:.2f}%")
```

```python
resumo = df.groupby("Class")["Amount"].agg(
    quantidade="count",
    media="mean",
    mediana="median"
)

resumo["percentual_ate_10"] = (
    df["Amount"].le(10)
    .groupby(df["Class"])
    .mean() * 100
)

display(resumo.round(2))
```

### Análise de transações de baixo valor

A análise identificou que 50,61% das transações fraudulentas possuem valor igual ou inferior a 10, contra 35,18% das transações legítimas.

A diferença de 15,43 pontos percentuais indica maior concentração proporcional de valores baixos entre as transações fraudulentas.

Esse resultado sugere que utilizar apenas limites elevados de valor como critério de prevenção pode deixar de identificar parte das fraudes.

Entretanto, o valor isoladamente não é suficiente para classificar uma transação como fraudulenta. São necessárias análises adicionais para avaliar sua capacidade de discriminar as classes.

```python
import matplotlib.pyplot as plt

percentuais = resumo["percentual_ate_10"].copy()
percentuais.index = ["Legítimas", "Fraudulentas"]

ax = percentuais.plot(
    kind="bar",
    figsize=(8, 5),
    color=["#64748B", "#2563EB"]
)

plt.title("Transações de valor até 10 por classe")
plt.xlabel("Tipo de transação")
plt.ylabel("Percentual (%)")
plt.xticks(rotation=0)
plt.ylim(0, 60)

for i, valor in enumerate(percentuais):
    ax.text(i, valor + 1, f"{valor:.2f}%", ha="center")

plt.tight_layout()
plt.show()
```

```python
fora_janela = df.loc[
    ~df["Time"].between(25198, 28755)
].copy()

alertas = fora_janela["V1"] < -15

verdadeiros_positivos = (
    alertas & (fora_janela["Class"] == 1)
).sum()

falsos_positivos = (
    alertas & (fora_janela["Class"] == 0)
).sum()

total_fraudes = (fora_janela["Class"] == 1).sum()

print("Transações sinalizadas:", alertas.sum())
print("Fraudes identificadas:", verdadeiros_positivos)
print("Legítimas sinalizadas:", falsos_positivos)
print(
    "Precisão dos alertas:",
    f"{verdadeiros_positivos / alertas.sum() * 100:.2f}%"
    if alertas.sum() else "Sem alertas"
)
print(
    "Fraudes detectadas:",
    f"{verdadeiros_positivos / total_fraudes * 100:.2f}%"
)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

y_real = fora_janela["Class"]
y_previsto = (fora_janela["V1"] < -15).astype(int)

matriz = confusion_matrix(
    y_real,
    y_previsto,
    labels=[0, 1]
)

visualizacao = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=["Legítima", "Fraude"]
)

visualizacao.plot(cmap="Blues", values_format="d")
plt.title("Matriz de Confusão — Regra V1 < -15")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Variáveis utilizadas pelo modelo
X = df.drop(columns=["Class"])

# Resultado que queremos prever
y = df["Class"]

# Separação dos dados
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Transações para treinamento:", len(X_treino))
print("Transações para teste:", len(X_teste))

print("\nFraudes no treinamento:", y_treino.sum())
print("Fraudes no teste:", y_teste.sum())
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

modelo = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    )
)

modelo.fit(X_treino, y_treino)

print("Modelo treinado com sucesso!")
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Fazer previsões nos dados de teste
y_predito = modelo.predict(X_teste)

# Calcular métricas
precisao = precision_score(y_teste, y_predito, zero_division=0)
recall = recall_score(y_teste, y_predito, zero_division=0)
f1 = f1_score(y_teste, y_predito, zero_division=0)

# Matriz de confusão
tn, fp, fn, tp = confusion_matrix(
    y_teste, y_predito, labels=[0, 1]
).ravel()

print("RESULTADOS DA REGRESSÃO LOGÍSTICA")
print(f"Precisão: {precisao:.2%}")
print(f"Recall: {recall:.2%}")
print(f"F1-score: {f1:.2%}")

print("\nMATRIZ DE CONFUSÃO")
print(f"Verdadeiros positivos: {tp}")
print(f"Falsos positivos: {fp}")
print(f"Falsos negativos: {fn}")
print(f"Verdadeiros negativos: {tn}")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
import pandas as pd

probabilidades = modelo.predict_proba(X_teste)[:, 1]

resultados = []

for limiar in [0.50, 0.70, 0.80, 0.90, 0.95]:
    previsoes = (probabilidades >= limiar).astype(int)

    resultados.append({
        "Limiar": limiar,
        "Alertas": int(previsoes.sum()),
        "Precisão": round(
            precision_score(y_teste, previsoes, zero_division=0) * 100, 2
        ),
        "Recall": round(
            recall_score(y_teste, previsoes, zero_division=0) * 100, 2
        ),
        "F1-score": round(
            f1_score(y_teste, previsoes, zero_division=0) * 100, 2
        )
    })

pd.DataFrame(resultados)
```

```python
from sklearn.model_selection import train_test_split

X_treino_novo, X_validacao, y_treino_novo, y_validacao = train_test_split(
    X_treino,
    y_treino,
    test_size=0.25,
    stratify=y_treino,
    random_state=42
)

print("Treinamento:", len(X_treino_novo))
print("Validação:", len(X_validacao))
print("Teste reservado:", len(X_teste))

print("\nFraudes no treinamento:", y_treino_novo.sum())
print("Fraudes na validação:", y_validacao.sum())
print("Fraudes no teste:", y_teste.sum())
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

modelo_v2 = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    )
)

modelo_v2.fit(X_treino_novo, y_treino_novo)

print("Modelo v2 treinado com sucesso!")
print("Transações utilizadas:", len(X_treino_novo))
print("Fraudes utilizadas:", y_treino_novo.sum())
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
import pandas as pd

# Pontuações calculadas no conjunto de validação
prob_validacao = modelo_v2.predict_proba(X_validacao)[:, 1]

resultados_validacao = []

for limiar in [0.50, 0.70, 0.80, 0.90, 0.95]:
    previsoes = (prob_validacao >= limiar).astype(int)

    resultados_validacao.append({
        "Limiar": limiar,
        "Alertas": int(previsoes.sum()),
        "Precisão (%)": round(
            precision_score(y_validacao, previsoes, zero_division=0) * 100, 2
        ),
        "Recall (%)": round(
            recall_score(y_validacao, previsoes, zero_division=0) * 100, 2
        ),
        "F1-score (%)": round(
            f1_score(y_validacao, previsoes, zero_division=0) * 100, 2
        )
    })

tabela_validacao = pd.DataFrame(resultados_validacao)

display(tabela_validacao)
```

```python
melhor_resultado = tabela_validacao.loc[
    tabela_validacao["F1-score (%)"].idxmax()
]

limiar_escolhido = float(melhor_resultado["Limiar"])

print(f"Limiar escolhido: {limiar_escolhido:.2f}")
print(f"Precisão na validação: {melhor_resultado['Precisão (%)']:.2f}%")
print(f"Recall na validação: {melhor_resultado['Recall (%)']:.2f}%")
print(f"F1-score na validação: {melhor_resultado['F1-score (%)']:.2f}%")
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Pontuações do conjunto de teste
prob_teste = modelo_v2.predict_proba(X_teste)[:, 1]

# Aplicar o limiar escolhido na validação
previsoes_teste = (prob_teste >= limiar_escolhido).astype(int)

# Calcular métricas
precisao_teste = precision_score(
    y_teste, previsoes_teste, zero_division=0
)

recall_teste = recall_score(
    y_teste, previsoes_teste, zero_division=0
)

f1_teste = f1_score(
    y_teste, previsoes_teste, zero_division=0
)

tn, fp, fn, tp = confusion_matrix(
    y_teste, previsoes_teste, labels=[0, 1]
).ravel()

print("AVALIAÇÃO DO MODELO V2")
print(f"Limiar: {limiar_escolhido:.2f}")
print(f"Precisão: {precisao_teste:.2%}")
print(f"Recall: {recall_teste:.2%}")
print(f"F1-score: {f1_teste:.2%}")

print("\nMATRIZ DE CONFUSÃO")
print(f"Fraudes identificadas: {tp}")
print(f"Falsos alertas: {fp}")
print(f"Fraudes não detectadas: {fn}")
print(f"Legítimas corretamente classificadas: {tn}")
```

```python
import pandas as pd

resumo_modelo = pd.DataFrame([{
    "modelo": "Regressão Logística v2",
    "limiar": limiar_escolhido,
    "precisao": round(precisao_teste, 4),
    "recall": round(recall_teste, 4),
    "f1_score": round(f1_teste, 4),
    "verdadeiros_positivos": int(tp),
    "falsos_positivos": int(fp),
    "falsos_negativos": int(fn),
    "verdadeiros_negativos": int(tn)
}])

display(resumo_modelo)
```

```python
import json
from pathlib import Path

destino = Path("../frontend/data/modelo_v2.json")

resultado = resumo_modelo.iloc[0].to_dict()

# Converter números do Pandas para tipos compatíveis com JSON
resultado = {
    chave: (
        int(valor)
        if chave in [
            "verdadeiros_positivos",
            "falsos_positivos",
            "falsos_negativos",
            "verdadeiros_negativos"
        ]
        else float(valor)
        if chave in ["limiar", "precisao", "recall", "f1_score"]
        else valor
    )
    for chave, valor in resultado.items()
}

resultado["tipo_avaliacao"] = "Experimento exploratório"
resultado["observacao"] = (
    "Conjunto de teste utilizado em explorações anteriores. "
    "Resultados não representam validação independente em produção."
)

destino.parent.mkdir(parents=True, exist_ok=True)

with destino.open("w", encoding="utf-8") as arquivo:
    json.dump(resultado, arquivo, indent=4, ensure_ascii=False)

print("Resultados exportados com sucesso!")
print("Arquivo:", destino.resolve())
```
